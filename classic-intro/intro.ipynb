```python
# Lets say we have 4 inputs and layer(ex 3 neurons) and each neuron have a bias associated with it so 3 bias values
# how many weights we need? well each input should interact with each neuron, so we have 4 inputs and 3 neurons --> 12 weights
# Lets represent mathematically.
```

```python
x = [2, 3, 4, 1]
W = [[0.2, 0.3, 0.4, 0.5], [0.3, 0.5, 0.6, 0.7], [0.2, 0.4, 0.1, 0.6]]
b = [0.2, 0.1, 0.6]
outputs = []
for i in range(0, len(W)):
    sum_value = 0
    for j in range(0, len(x)):
        sum_value += W[i][j]*x[j]
    sum_value += b[i]
    outputs.append(sum_value)
print(outputs)
```

```python
# Layer --> Mostly in our case it is input and weights, so vector @ matrix, order matters and the way we do matters
a = [1, 2, 3]
b = [[1, 2, 3], [2, 3, 4], [4, 5, 6], [1, 6, 7]]
a = np.array(a)
b = np.array(b)
```

```python
a @ b # --> will be a error as numpy will convert a to (1, 3) @ (4, 3) -->mismatch
```

```python
b @ a # this will be fine, because the numpy will convert the vector into a (4, 3) to (3, 1) --> (4, 1) -> 1 is droped -> (4,)
```

```python
# batch --> inputs are batched, matrix @ matrix
# Lets also consider the bias
a = [[2, 3, 4], [3, 5, 6], [3, 5, 6]] # --> inputs of length 3 stacked as a batch
b = [[1, 2, 3], [2, 3, 4], [4, 5, 6], [1, 6, 7]]
b = np.array(b)
a = np.array(a)
bias = [1, 2, 3, 4] # (1, outputs)
bias = np.array(bias).reshape(1, -1)
print(a)
print(b)
print(bias)
```
